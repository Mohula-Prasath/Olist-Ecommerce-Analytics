```python
import numpy as np
import pandas as pd
```

Preprocessing Olist_orders_dataset.csv

```python
df_orders=pd.read_csv("olist_orders_dataset.csv")
df_orders.head()
```

```python
df_orders.info()
```

```python
df_orders["order_purchase_timestamp"]=pd.to_datetime(df_orders["order_purchase_timestamp"])
df_orders["order_approved_at"]=pd.to_datetime(df_orders["order_approved_at"])
df_orders["order_delivered_carrier_date"]=pd.to_datetime(df_orders["order_delivered_carrier_date"])
df_orders["order_delivered_customer_date"]=pd.to_datetime(df_orders["order_delivered_customer_date"])
df_orders["order_estimated_delivery_date"]=pd.to_datetime(df_orders["order_estimated_delivery_date"])
```

```python
df_orders.info()
```

```python
df_orders.isnull().sum()
```

```python
df_orders=df_orders.drop_duplicates()
```

```python
df_orders.isnull().sum()
```

```python
df_orders_inconsistent=df_orders[(df_orders["order_status"]=="delivered") & ((df_orders["order_approved_at"].isna()) | 
                                                                             (df_orders["order_delivered_carrier_date"].isna()) |
                                                                             (df_orders["order_delivered_customer_date"].isna())) ]
print("number of rows: ",len(df_orders_inconsistent.index))
df_orders_inconsistent.index
```

```python
inconsistent_index=df_orders_inconsistent.index
```

```python
df_orders=df_orders.drop(index=inconsistent_index)
```

```python
df_orders.isnull().sum()
```

```python
df_orders["delivery_time"]=(df_orders["order_delivered_customer_date"]-df_orders["order_purchase_timestamp"]).dt.days
df_orders["approval_time"]=(df_orders["order_approved_at"]-df_orders["order_purchase_timestamp"]).dt.days
df_orders["carrier_time"]=(df_orders["order_delivered_customer_date"]-df_orders["order_delivered_carrier_date"]).dt.days
df_orders["expected_vs_actual_delivery"]=(df_orders["order_estimated_delivery_date"]-df_orders["order_delivered_customer_date"]).dt.days
```

```python
df_orders.info()
```

```python
df_orders.to_csv("preprocessed_olist_orders_dataset.csv",index=False)
```

Preprocessing olist_geolocation_dataset.csv

```python
df_geolocation=pd.read_csv("olist_geolocation_dataset.csv")
df_geolocation.head()
```

```python
df_geolocation.info()
```

```python
df_geolocation=df_geolocation.drop_duplicates()
df_geolocation.info()
```

```python
df_geolocation=df_geolocation.groupby(["geolocation_zip_code_prefix"]).agg({
    "geolocation_lat":"mean",
    "geolocation_lng": "mean",
    "geolocation_city": "first",
    "geolocation_state": "first"
}).reset_index()
df_geolocation.info()
```

```python
df_geolocation.to_csv("preprocessed_olist_geolocation_dataset.csv",index=False)
```
